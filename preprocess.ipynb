```python
import pandas as pd

pd.set_option("display.max_rows", 200)
pd.set_option("display.max_columns", 100)
```

```python
UR3e_raw_dir = r"energy-consumption-model-for-manipulators-main\UR3e\Raw Data"

UR3e_train_csv_files = [
    "UR_DataDynamics_record7.csv",
    "UR_DataDynamics_record8.csv",
    "UR_DataDynamics_record9.csv",
    "UR_DataDynamics_record11.csv",
    "UR_DataDynamics_record12.csv",
    "UR_DataDynamics_record13.csv",
    "UR3record14.csv",
    "UR3record15.csv",
    "UR3record16.csv",
    "UR3record17.csv",
    "UR3record18.csv",
    "UR3record19.csv",
    "UR3record20.csv",
    "UR3record21.csv"
]

UR3e_test_csv_files = ["UR3recordTesting2.csv", "UR3recordTesting4.csv"]

UR3e_train = pd.concat(
    [pd.read_csv(UR3e_raw_dir+"\\"+file).assign(source_file=file) for file in UR3e_train_csv_files],
    ignore_index=True
).set_index("t1").sort_index().reset_index()
UR3e_train["PC"] = UR3e_train["V"] * UR3e_train["I"]

UR3e_test = pd.concat(
    [pd.read_csv(UR3e_raw_dir+"\\"+file).assign(source_file=file) for file in UR3e_test_csv_files],
    ignore_index=True
).set_index("t1").sort_index().reset_index()
UR3e_test["PC"] = UR3e_test["V"] * UR3e_test["I"]

print("UR3e训练集：")
print(UR3e_train["source_file"].value_counts().sort_index())
display(UR3e_train)

print("UR3e测试集：")
print(UR3e_test["source_file"].value_counts().sort_index())
display(UR3e_test)
```

```python
def assign_trajectory_id(df, gap_threshold=1.0):
    """按 [source_file, ID] 分组划分轨迹。

    - 同一 source_file 下的同一个 ID 视为一组；
    - 组内按时间排序后，相邻时刻差 > gap_threshold（默认 1 s）即视为进入新轨迹；
    - trajectory_id 从 1 开始，按各段起始时刻从早到晚统一编号。
    """
    df = (
        df.sort_values(["source_file", "ID", "t1"], kind="mergesort")
        .reset_index(drop=True)
        .copy()
    )

    # 1) 组内分段：组内首行 diff 为 NaN（按 False 处理），之后差 > 阈值则段号 +1
    df["_seg"] = df.groupby(["source_file", "ID"], sort=False)["t1"].transform(
        lambda s: (s.diff() > gap_threshold).cumsum()
    )

    # 2) 每段的起始时刻，用于排序编号
    df["_start"] = df.groupby(["source_file", "ID", "_seg"])["t1"].transform("min")

    # 3) 按起始时刻从早到晚编号（同一时刻再按文件、ID 排序，保证结果可复现）
    order = (
        df[["source_file", "ID", "_seg", "_start"]]
        .drop_duplicates()
        .sort_values(["_start", "source_file", "ID"], kind="mergesort")
        .reset_index(drop=True)
    )
    order["trajectory_id"] = order.index + 1

    # 4) 回填编号，并按轨迹编号、时刻排序
    return (
        df.merge(order.drop(columns="_start"), on=["source_file", "ID", "_seg"], how="left")
        .drop(columns=["_seg", "_start"])
        .sort_values(["trajectory_id", "t1"], kind="mergesort")
        .reset_index(drop=True)
    )


def traj_summary(df):
    """轨迹概览：样本数、时间范围、最大/平均相邻间隔"""
    g = df.groupby(["trajectory_id", "source_file", "ID"])["t1"]
    return pd.DataFrame(
        {
            "样本数": g.size(),
            "起始时刻": g.min().round(2),
            "结束时刻": g.max().round(2),
            "最大相邻间隔": g.apply(lambda s: s.diff().max()).round(3),
            "平均相邻间隔": g.apply(lambda s: s.diff().mean()).round(4),
        }
    )


UR3e_train = assign_trajectory_id(UR3e_train)
UR3e_test = assign_trajectory_id(UR3e_test)

print(f"UR3e训练集：共划分出 {UR3e_train['trajectory_id'].nunique()} 条轨迹")
display(traj_summary(UR3e_train))

print(f"UR3e测试集：共划分出 {UR3e_test['trajectory_id'].nunique()} 条轨迹")
display(traj_summary(UR3e_test))
```

```python
UR3e_train.to_csv(r"data\UR3e_train.csv", index=False)
UR3e_test.to_csv(r"data\UR3e_test.csv", index=False)
```
